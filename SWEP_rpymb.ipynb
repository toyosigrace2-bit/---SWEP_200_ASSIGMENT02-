*LIST COMPREHENSION: ONE OF THE METHODS THAT ENSURES OPTIMIZED MACHINE LEARNING TRAINING*


```python
#extract the even numbers from the khan list
even_number_list = []
odd_number_list = list() #list() initializes a list
for i in khan:
  if i % 2 ==0:
    even_number_list.append(i) #append is a method that adds an element or value to an existing/new list
  else:

    odd_number_list.append(i)




#LIST COMPREHENSION
even_number_list2 = [i for i in khan if i % 2 ==0]
odd_number_list2 = [i for i in khan if i % 2 !=0]

print(even_number_list2)
print(odd_number_list2)
```

```python
class Swep():
  def __init__(self, uni, level, dept, count = 47): #number of students = 47 :instantiate the attributes of the class into the memory
    self.uni = uni
    self.level = level
    self.dept = dept
    self.count = count

  def product_of_student_count_with_lenght_of_uni(self): #method, belongs to a class
    product = self.count * self.level
    return product

uni = "OAU"
level = 250
dept = "CHE"

#STEP1: INITIALIZE THE CLASS
swep = Swep(uni, level, dept)

#STEP2: PERFORM THE ACTION: to perform an operation under a class, use the dot notation; for example: initialized class.method
swep.product_of_student_count_with_lenght_of_uni()
```
